```python
def analyze_engagement(likes_list):
    total_likes = 0

    for likes in likes_list:
        total_likes += likes

    print(f"Total Likes: {total_likes}")

    
    if total_likes >= 1000:
        print("Post Status: Viral Post 🚀")
    else:
        print("Post Status: Normal Engagement 👍")


post_likes = [200, 300, 150, 400]

analyze_engagement(post_likes)
```

```python
def check_medicine_stock(stock):
    print(f"Medicine Stock: {stock}")

    if stock < 10:
        print("Status: Low Stock Alert ⚠️")
    else:
        print("Status: Stock Sufficient ✅")

medicine_stock = 6
check_medicine_stock(medicine_stock)
```

```python
def check_rainfall(rainfall_data, required_level):
    total = 0

    for rain in rainfall_data:
        total += rain

    average = total / len(rainfall_data)

    print(f"Average Rainfall: {average:.0f}")

    if average >= required_level:
        print("Rainfall Status: Adequate Rainfall 🌧️✅")
    else:
        print("Rainfall Status: Inadequate Rainfall ⚠️")


season_rainfall = [60, 75, 80, 70, 65, 85]
required_minimum = 70

check_rainfall(season_rainfall, required_minimum)
```

```python
def detect_duplicates(usernames):
    unique_users = set(usernames)

    if len(unique_users) < len(usernames):
        print("Duplicate Accounts Found: Yes ⚠️")
    else:
        print("Duplicate Accounts Found: No ✅")


accounts = ["john123", "emma_01", "john123", "alex99"]

detect_duplicates(accounts)
```

```python
def filter_premium_crops(prices):
    premium_crops = []

    for price in prices:
        if price > 2000:
            premium_crops.append(price)

    return premium_crops

crop_prices = [1500, 2500, 1800, 3200, 1900]

result = filter_premium_crops(crop_prices)

print("Premium Crops:", result)
```

```python
def check_application_health(error_count):
    print("Error Count:", error_count)

    if error_count == 0:
        print("System Status: Healthy")
    elif error_count <= 5:
        print("System Status: Minor Issues")
    else:
        print("System Status: Critical Issues")


check_application_health(7)
```

```python
def check_transaction(amount):
    daily_limit = 50000
    
    print("Transaction Amount:", amount)
    
    if amount <= daily_limit:
        print("Transaction Status: Approved")
    else:
        print("Transaction Status: Rejected")

check_transaction(60000)
```

```python
def check_exam_eligibility(attendance_list):
    total_days = len(attendance_list)
    present_days = 0

    for day in attendance_list:
        if day == "P":
            present_days += 1

    attendance_percentage = (present_days / total_days) * 100

    print("Attendance Percentage:", attendance_percentage)

    if attendance_percentage >= 75:
        print("Exam Eligibility: Eligible")
    else:
        print("Exam Eligibility: Not Eligible")


attendance = ["P", "P", "A", "P", "P"]  # 4 out of 5 days present

check_exam_eligibility(attendance)
```

```python
def calculate_bill(units):
    bill = 0

    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = (100 * 3) + ((units - 100) * 5)
    else:
        bill = (100 * 3) + (100 * 5) + ((units - 200) * 7)

    if bill < 500:
        usage_status = "Low Usage"
    elif bill <= 1500:
        usage_status = "Moderate Usage"
    else:
        usage_status = "High Usage"

    return bill, usage_status


units_consumed = 250
total_bill, status = calculate_bill(units_consumed)

print("Units Consumed:", units_consumed)
print("Total Bill: ₹", total_bill)
print("Usage Level:", status)
```
