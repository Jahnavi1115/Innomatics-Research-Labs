# Problem Statement 1: Smart Parking Lot Management System

```python
def smart_parking(capacity, logs):
    current = 0
    peak = 0

    for log in logs:
        if log == "IN":
            current += 1
        elif log == "OUT":
            if current > 0:
                current -= 1

        if current > peak:
            peak = current

    print("Currently Parked Vehicles:", current)
    print("Peak Parking Usage:", peak)

    if current > capacity:
        print("Parking Status: OVER CAPACITY")
    else:
        print("Parking Status: Available")
```

```python
capacity = 50
logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]

smart_parking(capacity, logs)
```

# Problem Statement 2: Online Food Delivery Time Estimator

```python
def estimate_delivery_time(distance, traffic, weather):
    
    base_time = distance * 5
    
    if traffic == "Low":
        traffic_delay = 5
    elif traffic == "Medium":
        traffic_delay = 10
    elif traffic == "High":
        traffic_delay = 10
    else:
        traffic_delay = 0
    
    if weather == "Clear":
        weather_delay = 0
    elif weather == "Rainy":
        weather_delay = 5
    elif weather == "Stormy":
        weather_delay = 20
    else:
        weather_delay = 0
    
    total_time = base_time + traffic_delay + weather_delay
    
    print("Estimated Delivery Time:", total_time, "minutes")
```

```python
estimate_delivery_time(8, "High", "Rainy")
```

# Problem Statement 3: Movie Theatre Seat Occupancy Analyzer

```python
def seat_occupancy_analyzer(total_seats, booked_seats):
    
    booked_count = len(booked_seats)
    
    occupancy = (booked_count / total_seats) * 100
    
    print(f"Occupancy: {occupancy:.0f}%")
    
    if occupancy == 100:
        status = "Housefull"
    elif occupancy >= 75:
        status = "Almost Full"
    elif occupancy >= 40:
        status = "Moderate"
    else:
        status = "Low Occupancy"
    
    print("Show Status:", status)
    
    if occupancy >= 80:
        print("Suggestion: Consider Opening Additional Show")
```

```python
total_seats = 200
booked_seats = [1] * 150   # 150 seats booked

seat_occupancy_analyzer(total_seats, booked_seats)
```

# Problem Statement 4: Cloud Server Load Classification System


```python
def classify_server_load(cpu_readings):
    
    total = 0
    for value in cpu_readings:
        total += value
        
    average = total / len(cpu_readings)
    
    print(f"Average CPU Load: {average:.0f}%")
    
    if average < 50:
        status = "Normal"
    elif 50 <= average <= 80:
        status = "Warning"
    else:
        status = "Critical"
        
    print("Server Status:", status)
```

```python
cpu_readings = [45, 60, 70, 85, 90]

classify_server_load(cpu_readings)
```

#  Problem Statement 5: Smart Classroom Resource Usage Monitor


```python
def monitor_resource_usage(resource_usage):
    
    overused = []
    
    threshold = 8
    
    for resource, hours in resource_usage.items():
        if hours > threshold:
            overused.append(resource)
    
    if overused:
        print("Overused Resources:", ", ".join(overused))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")
```

```python
resource_data = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}

monitor_resource_usage(resource_data)
```

# Problem Statement 6: Online Event Registration Capacity Controller

```python
def manage_event_registration(capacity, registrations):
    
    confirmed = 0
    waitlisted = 0
    
    for i in range(registrations):
        if confirmed < capacity:
            confirmed += 1
        else:
            waitlisted += 1
    
    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlisted)
    
    if confirmed >= capacity:
        print("Registration Status: Closed")
    else:
        print("Registration Status: Open")
```

```python
manage_event_registration(100, 105)
```
