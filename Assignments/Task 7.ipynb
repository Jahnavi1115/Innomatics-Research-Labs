# Problem 1: Smart Light Controller

```python
class SmartLight:
    
    def __init__(self, name):
        self.name = name
        self.status = "OFF"   
    
    def turn_on(self):
        self.status = "ON"
    
    def turn_off(self):
        self.status = "OFF"
    
    def display_status(self):
        print(f"{self.name} is {self.status}")


light_name = "Bedroom Light"
action = "ON"

light = SmartLight(light_name)

if action == "ON":
    light.turn_on()
elif action == "OFF":
    light.turn_off()

light.display_status()
```

# Problem 2: Employee ID Card System

```python
class Employee:
    
    def __init__(self, name, emp_id, department):
        self.name = name
        self.emp_id = emp_id
        self.department = department
    
    def display_id_card(self):
        print("Employee ID Card")
        print("Name:", self.name)
        print("Employee ID:", self.emp_id)
        print("Department:", self.department)

employee_name = "Rahul"
employee_id = "EMP102"
department = "IT"

emp = Employee(employee_name, employee_id, department)

emp.display_id_card()
```

# Problem 3: Mobile Contact Record

```python
class Contact:
    
    def __init__(self, name, phone):
        self.name = name
        self.phone = phone
    
    def save_contact(self):
        print("Contact Saved")
        print("Name:", self.name)
        print("Phone:", self.phone)

contact_name = "Anita"
phone_number = "9876543210"

my_contact = Contact(contact_name, phone_number)

my_contact.save_contact()
```

# Problem 4: Product Price Tag Generator

```python
class Product:
    
    def __init__(self, name, price):
        self.name = name
        self.price = price
    
    def print_price_tag(self):
        print("Product:", self.name)
        print("Price: ₹{}".format(self.price))


product_name = "Headphones"
price = 2499

item = Product(product_name, price)

item.print_price_tag()
```

# Problem 5: Movie Rating Display System

```python
class Movie:
    
    def __init__(self, name, rating):
        self.name = name
        self.rating = rating
    
    def display_details(self):
        print("Movie:", self.name)
        print("Rating:", self.rating, "/ 5")


movie_name = "Inception"
rating = 4.8

film = Movie(movie_name, rating)

film.display_details()
```

# Problem 6: Delivery Address Manager

```python
class Delivery:
    
    def __init__(self, customer_name, address):
        self.customer_name = customer_name
        self.address = address
    
    def print_delivery_details(self):
        print("Delivery Details")
        print("Customer:", self.customer_name)
        print("Address:", self.address)


# 🔹 Sample Input
customer = "Suman"
address = "Hyderabad"

# Create object
order = Delivery(customer, address)

# Display delivery details
order.print_delivery_details()
```
